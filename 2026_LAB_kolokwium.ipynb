# Zadanie 1. Dwie kostki

Rzucamy dwiema sześciennymi kostkami. Niech $S$ oznacza sumę wyników na obu kostkach.

Wykonaj symulację `N = 10000` rzutów dwiema kostkami i odpowiedz na pytania:

1. Oszacuj prawdopodobieństwo zdarzenia $S \geq 10$.
2. Oszacuj prawdopodobieństwo zdarzenia, że $S$ jest liczbą parzystą.
4. Narysuj wykres (np. slupkowy) słupkowy liczby wystąpień poszczególnych sum.

Przydatne funkcje i operacje: `rng.choice`, `np.sum`, `np.arange`, `plt.bar`.

````python
N = 10000

# Rzuc N razy dwoma kostkami
rzuty = rng.choice(np.arange(1, 7), size=(N, 2), replace=True)

# Suma rzutow
suma = rzuty.sum(axis=1)

print(rzuty[:10])
print(suma[:10])
````

# Zadanie 2. Prosty spacer losowy
Rozważamy prosty spacer losowy. W każdym kroku pozycja zmienia się o $1$ albo o $-1$,
każda możliwość z prawdopodobieństwem $1/2$.
Dokładniej: niech $X_1,X_2,\ldots,X_n$ będą niezależnymi zmiennymi losowymi takimi, że
$$
 \mathbb{P}(X_i=-1)=\mathbb{P}(X_i=1)=\frac{1}{2},\quad i=1,2,\ldots,n.
$$
Wówczas spacer losowy $S_k, k=0,\ldots,n$, definiujemy następująco: $S_0=0$ oraz
$$
 S_i=X_1+X_2+\ldots+X_i,\quad i=1,2,\ldots,n.
$$

Wylosuj `num_walks = 50` spacerów losowych długości `n = 200`.
Każdy spacer powinien być zapisany w jednej kolumnie macierzy.

Wykonaj następujące polecenia:

1. Utwórz macierz kroków rozmiaru `(n, num_walks)`, zawierającą wartości `-1` oraz `1`.
2. Na podstawie kroków oblicz trajektorie spacerów losowych.
3. Narysuj na jednym wykresie wszystkie trajektorie.
4. Oszacuj prawdopodobieństwo tego, że w ostatnim kroku spacer losowy ma wartość dodatnią, czyli $\mathbb{P}(S_n>0)$.
5. Policz średnią końcową pozycję spośród `num_walks` spacerów, czyli wyestymuj $\mathbb{E}(S_n)$.

Przydatne funkcje i operacje: `rng.choice`, `np.cumsum`, `plt.plot`, `np.mean`, `np.sum`.

(W funkcji `np.cumsum` trzeba odpowiednio dobrać argument `axis`.)

````python
n = 100
num_walks = 20

# Wylosuj X_i dla wszystkich błądzeń
kroki = rng.choice([-1, 1], size=(n, num_walks), replace=True)

print(kroki.shape)
print(kroki[:5, :5])
````
